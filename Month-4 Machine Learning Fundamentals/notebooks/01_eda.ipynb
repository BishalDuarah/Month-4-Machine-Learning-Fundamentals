```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/telco_churn.csv")

df.head()
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
# Clean column names
df.columns = df.columns.str.strip()

# Drop leakage columns as are derived after churn happens → data leakage → unrealistic model performance.
df = df.drop(columns=["Churn Score", "Churn Reason"])

# Check shape
df.shape
```

```python
df["Churn Value"].value_counts(normalize=True)
```

```python
for col in df.columns:
    print(f"'{col}'")
```

```python
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace("  ", " ")
```

```python
print(df.columns.tolist())
```

```python
cols_to_drop = ["CustomerID", "Count", "Country", "State", "City",
                "Zip Code", "Lat Long", "Latitude", "Longitude"]

existing_cols = [col for col in cols_to_drop if col in df.columns]

df = df.drop(columns=existing_cols)
```

```python
df.shape
```

```python
df["Churn Value"].value_counts(normalize=True)
```

```python
df["Total Charges"] = pd.to_numeric(df["Total Charges"], errors="coerce")
df.isnull().sum()
```

```python
df[df["Total Charges"].isnull()][["Tenure Months", "Total Charges"]]
```

```python
df["Total Charges"] = df["Total Charges"].fillna(0)
```
